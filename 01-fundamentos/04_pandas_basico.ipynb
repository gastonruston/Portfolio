## Creación de un Data Frame

```python
pd.DataFrame({"Tabaco":[250,30], "No tabaco": [27,215],}) 

#Tabla. Columnas "Tabaco/No tabaco", filas (sin index) "0/1" Valores de columnas: 250/30, 27/215
```

```python
pd.DataFrame({"Tabaco":[250,30], "No tabaco": [27,215],}, index=["Cáncer","No cáncer"]) 

#index= da un nombre a las filas

#Tabla. Columnas "Tabaco/No tabaco", filas "Cáncer/No cáncer" Valores de columnas: 250/30, 27/215
```

## Creación de una serie. Uso de .index .dtype .values

```python
colores = ["rojo","blanco","negro","azul","violeta","verde","amarillo","celeste","rosa"]
serie_colores = pd.Series(colores)

#Salida: columna con índice, columna con la lista 'colores'
print(serie_colores)
print(serie_colores[3]) #Salida 'azul'. Devuelve el valor correspondiente al índice
print(serie_colores.dtype) #Salida 'str'. Devuelve el tipo de dato de la serie.
print(serie_colores.values) #Crea un arreglo NumPy con los datos de la serie y elimina el índice.
```

## Creación de un DataFrame a partir de un diccionario

```python
agenda = {"Nombre":["José","María","Javier","Gerardo","Mario"],"Teléfono":["2901785678","2901345565","2964113426","2901997896","2964156278"],"Dirección":["Casa1","Casa4","Casa2","Casa5","Casa3"]}
Tabla_agenda = pd.DataFrame(agenda)
print(Tabla_agenda)
```

Explicar diferencias entre: arreglo NumPy, data frame y serie:

*Arreglo NumPy: matriz de datos uni o bidimensional. Permite operaciones matemáticas. Devuelve valores solo por la posición, sin índice.

*DataFrame: crea una tabla con filas y columnas. Cada fila corresponde a un registro completo y tiene asignado un índice que puede manipularse. Permite cruzar información, filtrarla, combinarla y exportarla.

*Serie: Es una columna, una lista con un índice, seleccionada de una tabla.

## Selección

Selección de una serie

```python
jurisdiccion = defunciones["jurisdiccion"]
print(jurisdiccion)

```

Selección de un dataframe

```python
grupo_etario = defunciones[["grupo_etario"]]
print(grupo_etario)
print(grupo_etario.dtypes)

```

Los datos de tipo str permanecen en ese formato cuando se seleccionan como serie. Cuando se crea un DataFrame con ese tipo de datos son tratados como un objeto de Python (texto, números, listas, diccionarios). 

El manejo de datos nulos con object se realiza con np.nan o None y con str con pd.NA. 

"object" No detecta números mezclados con letras(ej: "30 años" y "30") y se producen errores silentes al ordenar datos o calcular frecuencias. Por esta razón también, los métodos de texto pueden arrojar NaN. Enlentece el procesamiento por que Python debe buscar celda por celda el tipo de objeto que contiene.

Selección de varias columnas

```python
defunciones[["jurisdiccion","cod_causa_muerte_CIE10"]]
```

Selección con iloc


```python
# Una fila
# #Selecciona la 17° fila
defunciones.iloc[16,:]
```

Selección con loc por etiquetas. 

```python
# Selecciona las columnas "sexo_nombre","grupo_causa_defuncion_CIE10" por etiqueta

defunciones = defunciones.set_index("jurisdiccion")
defunciones.loc[:,["sexo_nombre","grupo_causa_defuncion_CIE10"]]
```

#Filtrado de una variable numérica

iloc: incluye el inicio del rango y excluyye el final. No admite búsqueda por etiqueta.
loc: incluye el inicio y el final del rango. Puede indexar datos tipo str, admite búsqueda por etiqueta.

## Filtrado booleano


```python
# Filtra los registros de sexo masculino (sexo_id = 1)
defunciones[defunciones["sexo_id"]==1]

```

Filtrado por texto 

```python
#Filtra los códigos "A16"

defunciones[defunciones["cod_causa_muerte_CIE10"] == "A16"]
```

Filtrado por una palabra o fragmento de texto de un dato tipo str

```python
# Filtra las causas que contienen el fragmento "INFEC"

defunciones[defunciones["grupo_causa_defuncion_CIE10"].str.contains("INFEC")] # discrimina minúsculas y mayúsculas
```

Filtrado por condiciones con "&"

```python
# Filtra las defunciones de varones ocurridas en el segundo semestre del año

defunciones.loc[(defunciones.sexo_nombre == "masculino") & (defunciones.mes_def>6)]
```

Filtrado con condiciones con " | "

```python
# Filtra registros de defunciones por enfermedades infecciosas o por cáncer

defunciones.loc[(defunciones.grupo_causa_defuncion_CIE10.str.contains("INFEC"))|(defunciones.grupo_causa_defuncion_CIE10.str.contains("MALIG"))]
```
