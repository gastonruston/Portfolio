# Crear un arreglo del 0 al 9

```python
x=np.arange(10)
print (x)
```

# Arreglo de 5 valores equiestpaciados entre 0 y 10 

```python
x=np.linspace(0,10,num=5,endpoint=False,retstep=False,dtype=int,axis=0)
print (x)
```

# Arreglo con ceros

```python
x=np.zeros((3,4),dtype=int)
print (x)
```

# Arreglo con unos y reshape

# Filtrado booleano

```python
notas=np.array([2,4,6,6,6,7,8,10,9,7,7,8])
aprobados=notas[notas>=6]
print (aprobados)
```
