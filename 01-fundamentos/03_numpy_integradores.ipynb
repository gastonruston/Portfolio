## Generación de una muestra con distribución normal y medidas de resumen

```python
rng = np.random.default_rng(42)
muestra = rng.normal(50, 10, 1000)
print(f"Media: {muestra.mean()}")
print(f"Mediana: {np.median(muestra)}")
print(f"Desvío estándar poblacional: {muestra.std(ddof=0)}")
print(f"Desvío estándar muestral: {muestra.std(ddof=1)}")
print(f"Percentiles 25 y 75: {np.percentile(muestra,[25, 75])}")
```

"muestra" tiene 1000 valores, la media poblacional es 50, el DS poblacional es 10 y tiene distribución normal. Los estadísticos calculados de "muestra" guardan similitud a los poblacionales.

## Ventas por producto y trimestre. Creción de una matriz de 5×4 (5 productos, 4 trimestres) con ventas aleatorias. Cálculo del total por producto, total por trimestre y el producto con mayor venta total.

```python
#Semilla
rng=np.random.default_rng(42) 
#Matriz 5 x 4 de valores aleatorios. 5 productos 4 trimestres                                  
ventas = rng.integers (100, 1000, size = (5, 4))  
#Lista de productos               
productos = ["Laptop", "Mouse", "PC", "Impresora", "Micrófono"] 

#Calculo de totales por producto
        
total_laptop = np.sum(ventas[0,:])                  
total_mouse = np.sum(ventas[1,:])
total_pc = np.sum(ventas[2,:])
total_impresora = np.sum(ventas[3,:])
total_micrófono = np.sum(ventas[4,:])

#Cálculo de totales por trimestre

trimestre1 = np.sum(ventas[:,0])
trimestre2 = np.sum(ventas[:,1])
trimestre3 = np.sum(ventas[:,2])
trimestre4 = np.sum(ventas[:,3])

#Creación de un arreglo (5,) con el total de las ventas de cada producto

ventas_totales_por_producto = np.sum(ventas, axis =1)

#Búsqueda del índice con el valor máximo del arreglo

índice_más_vendido = np.argmax(ventas_totales_por_producto)

#Asignación del índice hallado a la lista de productos

producto_mas_vendido = productos[índice_más_vendido]


#Salidas

print(f"Ventas totales de laptop: {total_laptop}")
print(f"Ventas totales de mouse: {total_mouse}")
print(f"Ventas totales de pc: {total_pc}")
print(f"Ventas totales de impresora: {total_impresora}")
print(f"Ventas totales de micrófonos: {total_micrófono}")
print(f"Ventas totales del 1° trimestre: {trimestre1}")
print(f"Ventas totales del 2° trimestre: {trimestre2}")
print(f"Ventas totales del 3° trimestre: {trimestre3}")
print(f"Ventas totales del 4° trimestre: {trimestre4}")
print(f"Producto más vendido: {producto_mas_vendido}")
```
