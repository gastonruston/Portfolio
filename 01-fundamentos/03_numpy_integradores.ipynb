## Generación de una muestra con distribución normal y cálculo de las medidas de resumen

```python
rng = np.random.default_rng(42)
muestra = rng.normal(50, 10, 1000)
print(f"Media: {muestra.mean()}")
print(f"Mediana: {np.median(muestra)}")
print(f"Desvío estándar poblacional: {muestra.std(ddof=0)}")
print(f"Desvío estándar muestral: {muestra.std(ddof=1)}")
print(f"Percentiles 25 y 75: {np.percentile(muestra,[25, 75])}")
```

"muestra" tiene 1000 valores, la media poblacional es 50, el DS poblacional es 10 y tiene distribución normal. Los estadísticos calculados de "muestra" guardan similitud a los poblacionales. Media 49.71 frente a 50; desvío 9.89 frente a 10

## Ventas por producto y trimestre. Creación de una matriz de 5×4 (5 productos, 4 trimestres) con ventas aleatorias. Cálculo del total por producto, total por trimestre y el producto con mayor venta total.

```python
#Semilla
rng = np.random.default_rng(42)
#Creación de la matriz
ventas = rng.integers(100, 1000, size=(5, 4))
#Lista de productos
productos = ["Laptop", "Mouse", "PC", "Impresora", "Micrófono"]

#Cálculo de ventas por producto (axis=1) y trimestre (axis=0)
total_por_producto = ventas.sum(axis=1)
total_por_trimestre = ventas.sum(axis=0)

#np.argmax devuelve el índice correspondiente al valor máximo
indice_mas_vendido = np.argmax(total_por_producto)

#Salidas
print(ventas)
print("Total por producto:", total_por_producto)
print("Total por trimestre:", total_por_trimestre)
print("Producto más vendido:", productos[indice_mas_vendido])
```

axis=1 suma por producto y axis=0 suma por trimestre. La impresora es el producto más vendido y el trimestre 4 el de mayor venta.

## Z-score por columna. Normalizar cada columna de esa matriz con (x - media) / desvío, usando broadcasting y axis=0. 

```python
#Cálculo de promedios trimestrales

promedios_trimestres = ventas.mean(axis=0)   

#Cálculo de DS trimestrales
   
ds_trimestres = ventas.std(axis=0)  

#Cálculo del z-score. Broadcasting entre dos arreglos

z_score = (ventas-promedios_trimestres)/ds_trimestres

#Salidas

print(f"z-score: {np.round(z_score,2)}")
print(f"Medias normalizadas: {np.round(z_score.mean(axis=0),2)}")
print(f"Desvíos estándar normalizados: {np.round(z_score.std(axis=0),2)}")
```

Se calculó el z-score de cada valor trimestral y luego se comprobó la normalización al obtener las medias de 0 con DS de 1.

## Valores atípicos con RIC. Con los 1000 valores del ejercicio 1, calcular Q1, Q3 e IQR = Q3 - Q1. Detectar y contar los valores fuera de ese rango con filtrado booleano.

```python
#Cálculo de los percentiles y del RIC

q1 = np.percentile(muestra, 25)
q3 = np.percentile(muestra, 75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

mascara = (muestra < limite_inferior) | (muestra > limite_superior)
atipicos = muestra[mascara]

print(f"Q1: {q1:.2f}  Q3: {q3:.2f}  IQR: {iqr:.2f}")
print(f"Límites: {limite_inferior:.2f} a {limite_superior:.2f}")
print(f"Cantidad de atípicos: {mascara.sum()}")
print(atipicos)

```

Para datos normales, se esperan aproximadamente 7 atípicos de 1000. La cantidad calculada fue 9.

Desafío: bucle contra vectorizado. Con 10 000 valores, calcular la media con un bucle for y luego con np.mean. 
Medir los tiempos con %timeit en una celda cada uno.

```python

muestra_grande = rng.integers(0,10000, size = 10000)
print(np.mean(muestra_grande))
%timeit np.mean(muestra_grande)
```

```python
def calcular_media_bucle(muestra_grande):
    suma = 0.0
    for valor in muestra_grande:
        suma += valor
    return suma / len(muestra_grande)
promedio=calcular_media_bucle(muestra_grande)

%timeit calcular_media_bucle(muestra_grande)
print(promedio)
```

Conclusión: el bucle en C realizado en NumPy es más veloz que el bucle for de Python. La velocidad con np.mean es 12.3 mcs por loop, mientras que con un bucle for se extendió a 1.58 ms (NumPy es unas 130 veces más rápido). Además la sintaxis con NumPy se simplifica considerablemente.
