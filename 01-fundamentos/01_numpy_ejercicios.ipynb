## Un segmento es una vista, no una copia. 


```python
x = np.arange(10)
v = x[2:5]
v[0] = 99
print(x)      # x cambió: v es una vista de x

x = np.arange(10)
c = x[2:5].copy()
c[0] = 99
print(x)      # x no cambió: c es una copia independiente
```

Un segmento no copia los datos, apunta a los mismos. Si lo modifico, modifico el original. Para evitarlo uso .copy()

## Filtrado booleano (Devuelve  6 6 6 7 8 10 9 7 7 8)

```python
notas = np.array([2,4,6,6,6,7,8,10,9,7,7,8])
aprobados = notas[notas>=6]
print(aprobados)
```

**¿Qué diferencia hay entre np.arange(0, 10, 2) y np.linspace(0, 10, 5)?**

np.arange va de 0 a 10 sin incluirlo de 2 en 2. La salida es 0,2,4,6,8. arange devuelve el paso señalado en el argumento; linspace va de 0 a 10 y devuelve 5 valores equidistantes entre sí e incluye el 10. La salida es 0,2.5,5,7.5,10. 
linspace devuelve float por defecto y arange incluye el inicio pero excluye el final.

```python
print(np.arange(0,10,2))
print(np.linspace(0,10,5))
```

**¿Qué devuelve m[:, 2] si m es una matriz de 5×4?**

Devuelve la tercera columna: : toma todas las filas (5 valores) y el índice entero 2 elimina la dimensión de columnas, por eso el resultado es un array 1D de forma (5,).

```python
m = np.arange(20)
m = m.reshape(5,4)
print(m)
print(m[:,2])
```

**¿Por qué a[a > 5] no usa un bucle?**

a>5 crea un arreglo booleano tipo True/False. Este arreglo funciona como una máscara que se aplica a cada elemento y se seleccionan los que son True. El recorrido se realiza en C sobre memoria contigua y no en el intérprete de Python. No necesita escribirse un bucle y por eso es más rápido.
