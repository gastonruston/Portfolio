## Agregaciones

### Suma de los valores de un arreglo

sum devuelve la suma de los elementos contenidos en un arreglo

```python
rng = np.random.default_rng(42)
valores =  rng.random(100) 
print(np.sum(valores))
```

axis=0 colapsa las filas y da un valor por columna; axis=1 colapsa las columnas y da un valor por fila.

```python
rng = np.random.default_rng(42)
suma_columna = rng.random((3,4))
print(np.sum(suma_columna))
print(f"La suma de cada columna es: {np.sum(suma_columna,axis=0)}") #con axis=0, devuelve la suma de cada columna
```

### Mínimo - Máximo. Arreglos unidimensionales

```python
rng = np.random.default_rng(42)
p = rng.standard_cauchy(1000)
print(np.min(p))
print(np.max(p))

#segunda opción
print(p.min())
print(p.max())
```

### Mínimo - Máximo. Agregados multidimensionales

```python
rng = np.random.default_rng(42)
j = rng.integers(0,10,size=(3, 4))
print(j)
print(f"El mínimo de cada columna es {j.min(axis=0)}")
print(f"El máximo de cada fila es {j.max(axis=1)}")
```

Para limpiar datos:
  np.isinf(x)  Detecta valores infinitos en un arreglo. True = infinito
                                                        False= valores normales o NaN (Not a Number/Faltante)
  np.isfinite(x)  Detecta valores finitos en un arreglo. True = números reales limitados
                                                         False = infinito o NaN

```python
datos = (b/a)
datos_limpios = datos [np.isfinite(datos)]
print(datos_limpios) #Salida: 3. 2. 1.66666667 1.5
```

Incompatibilidad de broadcasting: Se comparan las formas de derecha a izquierda: 2 y 3 no coinciden y ninguno es 1.

```python
matriz1 = np.ones((3, 2))
matriz2 = np.arange(3)

try:
    print(matriz1 + matriz2)
except ValueError as e:
    print("Error:", e) # Error. No es posible el broadcasting con las formas (3,2) (3,)
```
