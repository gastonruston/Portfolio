# Operaciones con NumPy

## Suma de dos arreglos

```python
a = np.arange(5)
b = np.arange(2, 7)
#Salida: 2,4,6,8,10
print(a+b)
```

## Suma de un arreglo con un número fijo

```python
print(a+5)

suma = np.add(a, 5)
print (f"La suma es {suma}")

#Salida 5,6,7,8,9
```

## Resta de dos arreglos

```python
print(b-a) 

resta = np.subtract(b, a)
print (f"La resta es {resta}")

#Salida 2,2,2,2,2
```

## Multiplicación de dos arreglos

```python
print(a * b)

multiplicacion = np.multiply(a, b)
print(f"La multiplicación es {multiplicacion}")

# Salida 0,3,8,15,24
```

## División de dos arreglos

```python
x = np.arange(4, 13, 2)
y = np.arange(2, 11, 2)
print(x/y)

# Salida 2, 1.5, 1.333, 1.25, 1.2
```

## Potencia de dos arreglos

```python
print(a ** b)

potencia = np.power(a, b)
print(f"La potencia es {potencia}")

# Salida 1, 1, 16, 243, 4096
```

Pensé que 0**2= 1. Es un error; el resultado es 0.

## Raíz cuadrada

```python
h = np.array([16, 64, 25, 9])
raíz_cuadrada = np.sqrt(h)
print(f"La raíz cuadrada de h es {raíz_cuadrada}")

# Salida 4, 8, 5, 3
```

## Escalar de un arreglo. Potencia y multiplicación

```python
print(h * 2)  # 32 128 50 18
print(h ** 2) # 256 4096 625 81
```

## Agregaciones

## Suma de los valores de un arreglo

```python
l  =  np.random.random(100) 
print(np.sum(l))
```

```python
l = np.random.random((3, 4))
print(np.sum(l))
print(f"La suma de cada columna es: {np.sum(l,axis=0)}")
```

## Mínimo - Máximo. Arreglos unidimensionales

```python
rng = np.random.default_rng(42)
p = rng.standard_cauchy(1000)
print(np.min(p))
print(np.max(p))

#segunda opción
print(p.min())
print(p.max())
```

## Agregados multidimensionales

```python
rng = np.random.default_rng(42)
j = rng.integers(0,10,size=(3, 4))
print(j)
print(f"El mínimo de cada columna es {j.min(axis=0)}")
print(f"El máximo de cada fila es {j.max(axis=1)}")
```
