## Raíz cuadrada

```python
h = np.array([16, 64, 25, 9])
raíz_cuadrada = np.sqrt(h)
print(f"La raíz cuadrada de h es {raíz_cuadrada}")

# Salida 4, 8, 5, 3
```

## Agregaciones

### Suma de los valores de un arreglo

sum devuelve la suma de los elemntos contenidos en un arreglo

```python
rng = np.random.default_rng(42)
matriz =  np.random.random(100) 
print(np.sum(matriz))
```

axis=0 colapsa las filas y da un valor por columna; axis=1 colapsa las columnas y da un valor por fila.

```python
rng = np.random.default_rng(42)
matriz = np.random.random((3, 4))
print(np.sum(matriz))
print(f"La suma de cada columna es: {np.sum(matriz,axis=0)}") #con axis=0, devuelve la suma de cada columna
```

### Mínimo - Máximo. Arreglos unidimensionales

```python
rng = np.random.default_rng(42)
p = rng.standard_cauchy(1000)
print(np.min(p))
print(np.max(p))

#segunda opción
print(p.min())
print(p.max())
```

### Mínimo - Máximo. Agregados multidimensionales

```python
rng = np.random.default_rng(42)
j = rng.integers(0,10,size=(3, 4))
print(j)
print(f"El mínimo de cada columna es {j.min(axis=0)}")
print(f"El máximo de cada fila es {j.max(axis=1)}")
```
